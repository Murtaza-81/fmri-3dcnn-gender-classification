```python
import json
import numpy as np


def create_gender_labels(metadata_file):

    with open(metadata_file,'r') as f:
        metadata = json.load(f)

    labels = []

    for subject in metadata:

        sid = subject["subject_id"]

        if sid[0].lower() == "f":
            labels.append(0)
        else:
            labels.append(1)

    return np.array(labels)


y_train = create_gender_labels(
    "prepared_data/train_metadata.json"
)

y_test = create_gender_labels(
    "prepared_data/test_metadata.json"
)

np.save("prepared_data/y_train.npy", y_train)
np.save("prepared_data/y_test.npy", y_test)

print("Train labels:", y_train.shape)
print("Test labels:", y_test.shape)
```
